```python
import matplotlib.pyplot as plt
import xarray as xr

from daaf_plot.figure import InteractivePColorMesh
from daaf_plot.style import set_font_size
```

```python
set_font_size(8, 8, 8)
```

```python
ds = xr.open_dataset('data/lat_pressure_scores.nc').rename(
    interpolated_pressure='pressure',
    interpolated_latitude='latitude',
)
```

```python
plt.close('all')
```

```python
fig = InteractivePColorMesh(
    ds.lp_plot_diff.transpose(..., 'pressure', 'latitude'),
    x_dim='latitude',
    y_dim='pressure',
    facet_dim=None,
    colorbar_groups_dim=('stat_diff',),
    open_figure_kwargs={
        'figure_w': 8,
        'h_ratio_axes': 0.5,
        'pad_w_left': 1,
        'pad_w_right': 1,
        'pad_h_top': 0.5,
        'vcb_w': 0.15,
        'pad_w_ax_vcb': 0.3,
        'colorbar': 'vertical',
        'dpi': 75,
    },
    default_widget_values={
        'selection': {
            'truncation': '15',
            'stat_diff': 'rmse_diff',
            'level_variable': 'z',
            'step': 2,
        },
        'xy_scale': {
            'y_scale': 'linlogp',
            'y_inverted': True,
        },
        'colorbar': {
            'continuous_cmap_name': 'RdBu',
            'cmap_reverse': True,
            'scheme': 'divergent',
            'div_vcentre': 0,
            'div_vmax': 0.4,
        },
    },
)
```

```python
fig.interactive_show()
```
