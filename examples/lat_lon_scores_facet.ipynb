```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt
import xarray as xr

from daaf_plot.figure import InteractivePColorMesh
from daaf_plot.style import set_font_size
```

```python
set_font_size(8, 8, 8)
```

```python
ds = xr.open_dataset('data/latlon_scores.nc')
```

```python
plt.close('all')
```

```python
fig = InteractivePColorMesh(
    ds.level_latlon.transpose(..., 'latitude', 'longitude'),
    x_dim='longitude',
    y_dim='latitude',
    facet_dim='step',
    coastlines=True,
    open_figure_kwargs={
        'figure_w': 8,
        'num_rows': 2,
        'num_cols': 2,
        'h_ratio_axes': 0.5,
        'pad_w_left': 0.1,
        'pad_w_right': 0.5,
        'pad_h_in': 0.3,
        'pad_h_top': 0.7,
        'pad_h_bot': 0.1,
        'pad_w_in': 0.1,
        'vcb_w': 0.2,
        'pad_w_ax_vcb': 0.5,
        'colorbar': 'vertical',
        'dpi': 100,
        'projection': ccrs.PlateCarree(),
    },
    default_widget_values={
        'selection': {
            'pressure': 10,
        },
        'colorbar': {
            'scheme': 'sequential',
            'seq_vmin': 0,
            'seq_vmax': 200,
        },
    },
)
```

```python
fig.interactive_show()
```
